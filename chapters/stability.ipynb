# Stability analysis

We had to be careful when simulating a model with fixed points (like the SEIR model) because fixed points can imply that no further events are possible and the system is just stuck there. That's not always the case, as we can also have *dynamical equilibrium*: fixed points where the system still undergoes changes but where those events balance each other on average.

Regardless, the natural next step in systems with fixed points is to study the [**stability**](https://en.wikipedia.org/wiki/Stability_theory) of those fixed points.
