# Writing discrete models

Compartmental models provide a clear map of the variables needed to describe a system and of the mechanisms involved in their dynamics. In discrete time, we can follow a simple recipe to go from a compartmental model to a system of discrete equations tracking the change of key variables at every time step of the model.

We will track the expected values of quantities of interest over discrete. Say we care about some number or quantity $N(t)$, then we want to track its next value $N(t+1)$ as a function of the state of the system at time $t$. Easy enough.
